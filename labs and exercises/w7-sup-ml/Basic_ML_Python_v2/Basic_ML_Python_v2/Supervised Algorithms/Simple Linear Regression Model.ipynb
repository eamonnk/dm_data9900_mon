```python
#https://www.javatpoint.com/simple-linear-regression-in-machine-learning
import numpy as nm  
import matplotlib.pyplot as mtp  
import pandas as pd  
```

```python
data_set= pd.read_csv('DataSets/Salary_Data.csv') # Loading the dataset
```

```python
x = data_set.iloc[:, :-1].values  # The value of the second column
y = data_set.iloc[:, 1].values   # The first colum
```

```python
data_set
```

```python
# Splitting the dataset into training and test set.  
from sklearn.model_selection import train_test_split # Library = set of functions  
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size= 1/3, random_state=0)  
```

```python
#Fitting the Simple Linear Regression model to the training dataset  
from sklearn.linear_model import LinearRegression  
regressor= LinearRegression()  
regressor.fit(x_train, y_train) 
```

```python
#Prediction of Test and Training set result  
y_pred= regressor.predict(x_test)  
x_pred= regressor.predict(x_train)
```

```python
mtp.scatter(x_train, y_train, color="green")   
mtp.plot(x_train, x_pred, color="red")    
mtp.title("Salary vs Experience (Training Dataset)")  
mtp.xlabel("Years of Experience")  
mtp.ylabel("Salary(In Rupees)")  
mtp.show()   
```
